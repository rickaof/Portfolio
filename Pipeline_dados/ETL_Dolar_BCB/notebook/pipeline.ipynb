### 1.1 Conectar e extrair dados de uma API pública

```python
import requests

def extract_currency_data():
    url = "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/CotacaoDolarDia(dataCotacao=@dataCotacao)?@dataCotacao='06-10-2025'&$top=100&$format=json"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()["value"]
    else:
        raise Exception(f"API Error: {response.status_code}")
```

```python
def transform_currency_data(data):
    for item in data:
        item["dataHoraCotacao"] = item["dataHoraCotacao"].split("T")[0]
        item["tipoBoletim"] = item.get("tipoBoletim", "Desconhecido")
    return data
```
